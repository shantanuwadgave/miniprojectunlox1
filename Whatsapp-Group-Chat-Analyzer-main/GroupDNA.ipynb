```python
import numpy as np
from datetime import datetime
```

```python
with open("hostel_boys.txt","r",encoding="utf-8") as f:
    lines = f.readlines()
```

```python
messages=[]
sys_count = 0
med_count = 0
del_count = 0
emp_count = 0

for line in lines:
    line = line.strip()

    if line=="":
        emp_count+=1
        continue
    if " - " not in line:
        sys_count+=1
        continue
    timestamp,rest = line.split(" - ",1)

    if ": " not in rest:
        sys_count+=1
        continue

    sender,message = rest.split(": ",1)

    if message == "<Media omitted>":
        med_count+=1
    elif message == "This message was deleted":
        del_count+=1

    messages.append(
        {"timestamp":timestamp,
        "sender":sender,
        "text":message})

print("=" * 60)
print("CHAT PARSER SUMMARY")
print("=" * 60)

print(f"Total Parsed Messages : {len(messages)}")
print(f"System Messages       : {sys_count}")
print(f"Media Messages        : {med_count}")
print(f"Deleted Messages      : {del_count}")
print(f"Empty Lines           : {emp_count}")

print("\nFirst 5 Messages:\n")

for msg in messages[:5]:
    print(msg)

print("\nLast 5 Messages:\n")

for msg in messages[-5:]:
    print(msg)
```

```python
def get_symbol(value, maximum):

    if maximum == 0:
        return "."

    percentage = value / maximum

    if percentage <= 0.25:
        return "."
    elif percentage <= 0.50:
        return "░"
    elif percentage <= 0.75:
        return "▒"
    else:
        return "█"
```

```python
stop_words = {
    "i", "is", "the", "a", "an", "and", "or", "to",
    "of", "in", "on", "for", "at", "with", "it",
    "this", "that", "you", "me", "my", "we", "our",
    "your", "are", "am", "was", "were", "be", "have",
    "has", "had", "do", "does", "did"
}


word_count = {}
for msg in messages:
    text = msg["text"]

    if text == "<Media omitted>":
        continue
    elif text == "This message was deleted":
        continue

    text = text.lower()

    punctuation = ".,!?;:()[]{}\"'`<>/@#$%^&*-_=+|\\"

    for ch in punctuation:
        text = text.replace(ch,"")

    words = text.split()

    for word in words:
        if word in word_count:
            word_count[word]+=1
        else:
            word_count[word]=1
    
    
    sorted_words = sorted(
    word_count.items(),
    key=lambda x: x[1],
    reverse=True
)
```

```python
print("=" * 60)
print("TOP 10 WORDS".center(60))
print("=" * 60)

for word, count in sorted_words[:10]:
    print(f"{word:<15} : {count}")
```

```python
from datetime import datetime
response_times = {}

for i in range(1,len(messages)):
    previous = messages[i-1]
    current = messages[i]

    if previous["sender"] == current["sender"]:
        continue
    t1 = datetime.strptime(previous["timestamp"], "%d/%m/%y, %H:%M")
    t2 = datetime.strptime(current["timestamp"], "%d/%m/%y, %H:%M")

    gap = (t2 - t1).total_seconds()/60

    sender = current["sender"]


    if sender not in response_times:
        response_times[sender] = []

    response_times[sender].append(gap)
```

```python
average_response = {}
for person,gaps in response_times.items():
    average_response[person] = sum(gaps) / len(gaps)
```

```python
print("=" * 60)
print("RESPONSE SPEED".center(60))
print("=" * 60)

for person, avg in sorted(average_response.items(), key=lambda x: x[1]):
    print(f"{person:<10} : {avg:.2f} minutes")
```

```python
all_dates = sorted({
    datetime.strptime(msg["timestamp"].split(",")[0], "%d/%m/%y").date()
    for msg in messages
})

activity = {}

for person in people:
    activity[person] = set()

for msg in messages:

    person = msg["sender"]

    date = datetime.strptime(
        msg["timestamp"].split(",")[0],
        "%d/%m/%y"
    ).date()

    activity[person].add(date)

silent_streak = {}

for person in people:

    longest = 0
    current = 0

    for day in all_dates:

        if day in activity[person]:
            current = 0
        else:
            current += 1

            if current > longest:
                longest = current

    silent_streak[person] = longest
```

```python
print("=" * 60)
print("LONGEST SILENT STREAK".center(60))
print("=" * 60)

for person, days in sorted(silent_streak.items(), key=lambda x: x[1], reverse=True):

    print(f"{person:<10} : {days} days")
```

```python
print("=" * 60)
print("RESPONSE PATTERNS".center(60))
print("=" * 60)

fastest = min(average_response, key=average_response.get)
slowest = max(average_response, key=average_response.get)

print(f"Fastest Replier : {fastest} ({average_response[fastest]:.2f} minutes)")
print(f"Slowest Replier : {slowest} ({average_response[slowest]:.2f} minutes)")

print("\nLongest Silent Streaks\n")

for person, days in sorted(silent_streak.items(), key=lambda x: x[1], reverse=True):
    print(f"{person:<10} : {days} days")
```

```python
#The night owl
night_score = {}

for person in people:

    total = 0
    night = 0

    for msg in messages:

        if msg["sender"] != person:
            continue

        total += 1

        hour = int(msg["timestamp"].split(", ")[1].split(":")[0])

        if hour >= 23 or hour <= 4:
            night += 1

    if total == 0:
        night_score[person] = 0
    else:
        night_score[person] = (night / total) * 100
```

```python
#The story teller
story_score = {}

for person in people:

    total_words = 0
    total_messages = 0

    for msg in messages:

        if msg["sender"] == person:

            total_messages += 1
            total_words += len(msg["text"].split())

    if total_messages == 0:
        story_score[person] = 0
    else:
        story_score[person] = total_words / total_messages
```

```python
#Drama Queen
drama_score = {}

for person in people:

    drama = 0
    total = 0

    for msg in messages:

        if msg["sender"] != person:
            continue

        total += 1

        text = msg["text"]

        if len(text) > 2:

            if text.isupper() or text.count("!") >= 2:
                drama += 1

    if total == 0:
        drama_score[person] = 0
    else:
        drama_score[person] = (drama / total) * 100
```

```python
#Group mom
caring_words = [
    "okay","safe","eat","sleep","please",
    "take","care","reminder","drink","water",
    "dont","forget","are","you"
]

mom_score = {}

for person in people:

    score = 0

    for msg in messages:

        if msg["sender"] != person:
            continue

        text = msg["text"].lower()

        for word in caring_words:

            if word in text:
                score += 1

    mom_score[person] = score
```

```python
#Comedian
funny = ["lol","haha","lmao","rofl","lmfao"]

comedy_score = {}

for person in people:

    total = 0
    funny_msgs = 0

    for msg in messages:

        if msg["sender"] != person:
            continue

        total += 1

        text = msg["text"].lower()

        for word in funny:

            if word in text:
                funny_msgs += 1
                break

    if total == 0:
        comedy_score[person] = 0
    else:
        comedy_score[person] = funny_msgs / total
```

```python
#Question master
question_score = {}

for person in people:

    total = 0
    questions = 0

    for msg in messages:

        if msg["sender"] != person:
            continue

        total += 1

        if msg["text"].strip().endswith("?"):
            questions += 1

    if total == 0:
        question_score[person] = 0
    else:
        question_score[person] = questions / total
```

```python
#Spammer
burst = {}

for person in people:
    burst[person] = []

current_sender = messages[0]["sender"]
count = 1

for i in range(1, len(messages)):

    if messages[i]["sender"] == current_sender:
        count += 1
    else:
        burst[current_sender].append(count)

        current_sender = messages[i]["sender"]
        count = 1

burst[current_sender].append(count)

spam_score = {}

for person in people:

    if len(burst[person]) == 0:
        spam_score[person] = 0
    else:
        spam_score[person] = sum(burst[person]) / len(burst[person])
```

```python
print("=" * 70)
print("PERSONALITY ARCHETYPES".center(70))
print("=" * 70)

for person in people:

    scores = {
        "THE SPAMMER": spam_score[person],
        "THE GROUP MOM": mom_score[person],
        "THE NIGHT OWL": night_score[person],
        "THE STORYTELLER": story_score[person],
        "THE DRAMA QUEEN": drama_score[person],
        "THE GHOST": ghost_score[person],
        "THE COMEDIAN": comedy_score[person],
        "THE QUESTION MASTER": question_score[person]
    }

    best = max(scores, key=scores.get)

    print(f"{person:<10} → {best}")
```
